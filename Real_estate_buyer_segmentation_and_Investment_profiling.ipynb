```python
# REAL ESTATE BUYER SEGMENTATION
# Machine Learning Based Buyer Segmentation and Investment Profiling

import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Sample real-estate buyer data
data = {
    "Buyer": ["A","B","C","D","E","F","G","H","I","J"],
    "Age": [25, 30, 45, 50, 28, 35, 55, 40, 32, 48],
    "Income_Lakh": [4, 5, 12, 15, 6, 8, 18, 10, 7, 14],
    "Budget_Lakh": [30, 40, 100, 150, 45, 70, 200, 90, 55, 130]
}

df = pd.DataFrame(data)

# Machine Learning: K-Means clustering
X = df[["Income_Lakh", "Budget_Lakh"]]

model = KMeans(n_clusters=3, random_state=42, n_init=10)
df["Buyer_Segment"] = model.fit_predict(X)

# Investment profile
profiles = {
    0: "Budget Buyer",
    1: "Mid-Range Investor",
    2: "Premium Investor"
}

df["Investment_Profile"] = df["Buyer_Segment"].map(profiles)

print("REAL ESTATE BUYER SEGMENTATION")
print(df)

# Visualization
plt.figure(figsize=(8,5))
plt.scatter(df["Income_Lakh"], df["Budget_Lakh"])
plt.xlabel("Income (Lakh)")
plt.ylabel("Property Budget (Lakh)")
plt.title("Real Estate Buyer Segmentation")
plt.show()
```

Machine Learning Based Buyer Segmentation and Investment Profiling for Real Estate Market Intelligence

1. Introduction

2. Objective

The main objectives are:

- To identify buyer investment profiles.
- To understand the relationship between income and property budget.
- To support better real-estate market analysis.

A sample dataset containing buyer age, income and property budget is used for analysis.

4. Methodology

The project uses the K-Means clustering machine learning algorithm. K-Means groups buyers with similar income and property-budget characteristics into clusters.

5. Buyer Segmentation

The buyers are divided into three groups:

- Budget Buyer
- Mid-Range Investor
- Premium Investor

6. Results

The graph shows the relationship between buyer income and property budget. The clustering model helps identify different categories of real-estate buyers.

7. Conclusion

Machine learning can help real-estate businesses understand different buyer groups. Buyer segmentation and investment profiling can support better market intelligence and decision-making.

```python
# Buyer Segment Summary

segment_counts = df["Investment_Profile"].value_counts()

plt.figure(figsize=(7,5))
segment_counts.plot(kind="bar")
plt.xlabel("Investment Profile")
plt.ylabel("Number of Buyers")
plt.title("Number of Buyers in Each Investment Profile")
plt.xticks(rotation=0)
plt.show()
```
